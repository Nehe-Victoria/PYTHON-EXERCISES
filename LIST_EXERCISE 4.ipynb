```python
# Sorting the list
ages = [19, 22, 19, 24, 20, 25, 26, 24, 25, 24]
ages.sort()
print(ages)
```

```python
# Finding the minimum and maximum ages
min_ages=min(ages)
max_ages=max(ages)
print("The minimum age is:",min_ages)
print("The minimum age is:",max_ages)
```

```python
# Add the min age and the max age again to the list
ages.append(min_ages)
ages.append(max_ages)
print("the updated list is:",ages)
```

```python
# Find the median age (one middle item or two middle items divided by two)
ages.sort()
n=len(ages)
if n%2==1:
        median=ages[n//2]
else:
    median=(ages[n//2-1]+ages[n//2])/2
    print(median)
```

```python
# Find the average age (sum of all items divided by their number )
total_age=sum(ages)
num_age=len(ages)
average_age=total_age/num_age
print(average_age)
```

```python
# Find the range of the ages (max minus min)
range= max_age-min_age
print("The range is:",range)
```

```python
# Compare the value of (min - average) and (max - average), use abs() method
min_diff=abs(min_age-average_age)
max_diff=abs(max_age-average_age)
print("The minimum diffrence is:",min_diff)
print("The maximum diffrence is:",max_diff)
```
